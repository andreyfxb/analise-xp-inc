```python
# Importa o pandas para trabalhar com tabelas de dados
import pandas as pd

# Importa o yfinance para buscar dados financeiros
import yfinance as yf
```

```python
# Solicita o histórico da ação XP a partir de 1º de dezembro de 2019
# Guarda os dados na variável xp, sem salvar um arquivo no computador
xp = yf.download("XP", start="2019-12-01")

# Mostra as cinco primeiras linhas para conferir o resultado
xp.head()
```

```python
# Mantém apenas o primeiro nível do cabeçalho, com os nomes dos campos
xp.columns = xp.columns.get_level_values(0)

# Mostra as cinco primeiras linhas para conferir a mudança
xp.head()
```

```python
# Mostra a quantidade de linhas e colunas, nessa ordem
xp.shape
```

```python
# Identifica valores ausentes e conta quantos existem em cada coluna
# Não verifica se os números preenchidos estão corretos
xp.isna().sum()
```

```python
# Mostra as cinco últimas linhas para observar datas, preços e volumes
xp['Close'].tail()
```

```python
# Recalcula as variações com os fechamentos que estão agora na variável xp
variacao_diaria = xp["Close"].pct_change() * 100

# Mostra as cinco últimas variações para conferir o resultado
variacao_diaria.tail()
```

```python
# Calcula a variação percentual entre os dois fechamentos informados
((15.870 / 15.960) - 1) * 100
```

```python
# Cria uma tabela para comparar os fechamentos com a variação recalculada
conferencia = pd.DataFrame({
    "Fechamento": xp["Close"],
    "Variacao_%": xp["Close"].pct_change() * 100
})

# Mostra os cinco últimos registros da conferência
conferencia.tail()
```

```python
# Cria uma cópia com os registros anteriores a 28/10/2026
xp_fechado = xp.loc[xp.index < "2026-10-28"].copy()

# Mostra as últimas cinco linhas para conferir o corte
xp_fechado.tail()
```

### Tratamento do pregão em andamento

- Data da conferência: 28/10/2026, às 12h16, horário de Brasília.
- O pregão ainda estava em andamento; os dados do dia poderiam mudar.
- Criamos a tabela xp_fechado somente com datas anteriores a 28/10/2026.
- A tabela original xp foi preservada.
- As próximas análises usarão xp_fechado.

```python
# Mostra a quantidade de linhas e colunas da base com pregões encerrados
xp_fechado.shape
```

```python
# Importa a ferramenta Path para trabalhar com caminhos de arquivos e pastas
from pathlib import Path

# Mostra a pasta de trabalho atual
Path.cwd()
```

```python
# Monta o caminho: sobe uma pasta e entra em data
pasta_dados = Path.cwd().parent / "data"

# Verifica se esse caminho corresponde a uma pasta existente
pasta_dados.is_dir()
```

```python
# Define o nome e o caminho do arquivo CSV
arquivo_csv = pasta_dados / "xp_historico_fechado.csv"

# Salva a tabela, incluindo as datas que estão no índice
xp_fechado.to_csv(arquivo_csv, index=True)

# Mostra onde o arquivo foi salvo
arquivo_csv
```

```python
# Verifica se o arquivo CSV existe no caminho indicado
arquivo_csv.is_file()
```

```python
# Lê o CSV e usa a coluna Date como índice
# Converte as datas de texto para o tipo de data do pandas
xp_conferencia = pd.read_csv(
    arquivo_csv,
    index_col="Date",
    parse_dates=True
)

# Mostra a quantidade de linhas e colunas do arquivo lido
xp_conferencia.shape
```

```python
# Compara a tabela original com a tabela lida do CSV
# Aceita diferenças numéricas muito pequenas causadas pela gravação e leitura
pd.testing.assert_frame_equal(
    xp_fechado,
    xp_conferencia,
    check_names=False,
    check_freq=False,
    rtol=1e-9,
    atol=1e-9
)
```
